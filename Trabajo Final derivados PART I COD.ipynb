```python
"""
Simulación Monte Carlo de la TRM (COP por USD) a 40 trimestres y conversión
de las cuotas del crédito en dólares a pesos
"""
import numpy as np
import pandas as pd

# ----------------------------- PARÁMETROS ---------------------------------
S0 = 3157.94            # TRM del 27 oct 2026 (Banco de la República, serie 1 de suameca)
FECHA_INICIO = "2026-10-27"
N_Q = 40                # trimestres (10 años)
N_SIM = 20000
SEED = 42

# Escenarios al trimestre 4: (nombre, probabilidad, límite inferior, límite superior)
ESCENARIOS = [
    ("Base",        0.50, 3250.0, 3500.0),
    ("Peso fuerte", 0.25, 3000.0, 3250.0),
    ("Peso débil",  0.25, 3500.0, 3900.0),
]
Z80 = 1.2816            # el rango del escenario = intervalo de 80 % de la lognormal
VOL_ANIO1 = 0.10        # volatilidad anual del puente dentro del año 1 (ruido intra-año)
MOSTRAR_TABLAS = True   # imprime los valores proyectados de los 40 trimestres
MOSTRAR_GRAFICAS = True # dibuja las gráficas (en Colab/Jupyter aparecen debajo de la celda)

# Datos históricos: TRM de cierre de cada trimestre (marzo, junio, septiembre, diciembre)
ARCHIVO_HIST = "trm_historica_trimestral.csv"
# Ventanas: los rendimientos usados son los posteriores a la fecha indicada
# y hasta el último dato (septiembre de 2026).
VENTANAS = {
    "3 años (2023-2026)":           "2023-09-30",
    "5 años (2021-2026)":           "2021-09-30",
    "10 años (2016-2026)":          "2016-09-30",
    "20 años (2006-2026)":          "2006-09-30",
    "Flotación libre (1999-2026)":  "1999-09-30",
    "Serie completa (1993-2026)":   "1993-03-31",
}
VENTANA_ELEGIDA = "Flotación libre (1999-2026)"

# Serie histórica embebida (TRM de cierre de cada trimestre). Permite correr el código
# sin el archivo CSV, por ejemplo en Google Colab. Fuentes: Superfinanciera (1993-2022),
# FRED/OCDE COLCCUSSP02STQ (2023), Pound Sterling Live y ValutaFX, cierres de mercado
# (2024-2026), y TRM del 31-dic-2024 (Superfinanciera).
TRM_HISTORICA_CSV = """fecha,trm
1993-03-31,767.10
1993-06-30,786.80
1993-09-30,810.11
1993-12-31,804.04
1994-03-31,821.21
1994-06-30,820.47
1994-09-30,840.90
1994-12-31,829.32
1995-03-31,873.67
1995-06-30,876.89
1995-09-30,977.54
1995-12-31,986.75
1996-03-31,1048.62
1996-06-30,1070.07
1996-09-30,1032.29
1996-12-31,1003.98
1997-03-31,1060.08
1997-06-30,1087.86
1997-09-30,1247.69
1997-12-31,1287.12
1998-03-31,1362.87
1998-06-30,1375.53
1998-09-30,1554.09
1998-12-31,1510.99
1999-03-31,1532.94
1999-06-30,1735.82
1999-09-30,2004.97
1999-12-31,1872.12
2000-03-31,1954.04
2000-06-30,2132.82
2000-09-30,2222.37
2000-12-31,2186.21
2001-03-31,2278.78
2001-06-30,2305.66
2001-09-30,2328.23
2001-12-31,2306.90
2002-03-31,2282.33
2002-06-30,2364.25
2002-09-30,2751.23
2002-12-31,2814.89
2003-03-31,2959.01
2003-06-30,2826.95
2003-09-30,2840.08
2003-12-31,2807.20
2004-03-31,2682.09
2004-06-30,2694.09
2004-09-30,2608.30
2004-12-31,2389.75
2005-03-31,2363.23
2005-06-30,2324.22
2005-09-30,2288.22
2005-12-31,2284.22
2006-03-31,2293.38
2006-06-30,2579.08
2006-09-30,2394.31
2006-12-31,2238.79
2007-03-31,2190.30
2007-06-30,1960.61
2007-09-30,2023.19
2007-12-31,2014.76
2008-03-31,1834.96
2008-06-30,1923.02
2008-09-30,2184.76
2008-12-31,2243.59
2009-03-31,2544.24
2009-06-30,2145.21
2009-09-30,1925.49
2009-12-31,2044.23
2010-03-31,1921.88
2010-06-30,1913.15
2010-09-30,1801.01
2010-12-31,1913.98
2011-03-31,1870.60
2011-06-30,1772.32
2011-09-30,1929.01
2011-12-31,1942.70
2012-03-31,1792.07
2012-06-30,1784.60
2012-09-30,1800.52
2012-12-31,1768.23
2013-03-31,1832.20
2013-06-30,1929.00
2013-09-30,1908.29
2013-12-31,1926.83
2014-03-31,1969.45
2014-06-30,1881.19
2014-09-30,2022.00
2014-12-31,2392.46
2015-03-31,2598.36
2015-06-30,2598.68
2015-09-30,3086.75
2015-12-31,3149.47
2016-03-31,3000.63
2016-06-30,2919.01
2016-09-30,2880.08
2016-12-31,3000.71
2017-03-31,2885.57
2017-06-30,3050.43
2017-09-30,2936.67
2017-12-31,2984.00
2018-03-31,2780.47
2018-06-30,2930.80
2018-09-30,2972.18
2018-12-31,3249.75
2019-03-31,3174.79
2019-06-30,3205.67
2019-09-30,3477.45
2019-12-31,3277.14
2020-03-31,4054.54
2020-06-30,3756.28
2020-09-30,3865.47
2020-12-31,3432.50
2021-03-31,3678.62
2021-06-30,3748.50
2021-09-30,3812.77
2021-12-31,3981.16
2022-03-31,3756.03
2022-06-30,4151.21
2022-09-30,4590.54
2022-12-31,4810.20
2023-03-31,4627.27
2023-06-30,4191.28
2023-09-30,4053.76
2023-12-31,3822.05
2024-03-31,3858.50
2024-06-30,4149.30
2024-09-30,4190.90
2024-12-31,4409.15
2025-03-31,4185.60
2025-06-30,4057.20
2025-09-30,3925.20
2025-12-31,3769.50
2026-03-31,3642.80
2026-06-30,3446.70
2026-09-30,3346.70
"""


def cargar_historico(archivo=ARCHIVO_HIST):
    """Lee el CSV si está en la carpeta de trabajo; si no, usa la serie embebida."""
    import io, os
    if archivo and os.path.exists(archivo):
        h = pd.read_csv(archivo, parse_dates=["fecha"])
    else:
        h = pd.read_csv(io.StringIO(TRM_HISTORICA_CSV), parse_dates=["fecha"])
    return h.sort_values("fecha")

# Crédito (punto A.2): SBA 7(a) a tasa fija, sistema francés trimestral
VALOR_MAQUINARIA_COP = 400_000_000
APORTE = 0.10
TASA_NOMINAL = 0.13     # Prime 7,00 % + 6,0 pp (tope SBA 7(a) USD 50.001-250.000)
PAGOS_ANIO = 4
# --------------------------------------------------------------------------


def estimar_parametros(fecha_desde, archivo=ARCHIVO_HIST):
    """Deriva y volatilidad trimestrales (en logaritmos) desde los datos históricos."""
    h = cargar_historico(archivo)
    r = np.log(h["trm"]).diff()
    r.index = h["fecha"]
    r = r.dropna()
    r = r[r.index > pd.Timestamp(fecha_desde)]
    mu_q = r.mean()
    sig_q = r.std(ddof=1)
    return {
        "n_trimestres": len(r),
        "desde": r.index.min().date(), "hasta": r.index.max().date(),
        "mu_q": mu_q, "sig_q": sig_q,
        "deriva_anual": 4 * mu_q,
        "volatilidad_anual": 2 * sig_q,
        "error_est_deriva_anual": 2 * sig_q / np.sqrt(len(r) / 4),
    }


def mediana_y_sigma(lo, hi):
    """Mediana geométrica y sigma log tal que [lo, hi] sea el intervalo de 80 %."""
    med = np.sqrt(lo * hi)
    sig = np.log(hi / lo) / (2 * Z80)
    return med, sig


def trayectoria_escenario_mediana(med, mu_q):
    """Camino mediano: interpolación log-lineal en el año 1; luego deriva histórica."""
    t = np.arange(N_Q + 1)
    logp = np.empty(N_Q + 1)
    logp[:5] = np.log(S0) + (np.log(med) - np.log(S0)) * t[:5] / 4
    for k in range(5, N_Q + 1):
        logp[k] = logp[k - 1] + mu_q
    return np.exp(logp)


def estimar_nu(fecha_desde, archivo=ARCHIVO_HIST):
    """Grados de libertad de una t de Student ajustada por máxima verosimilitud
    a los rendimientos log trimestrales de la ventana histórica."""
    from scipy import stats
    h = cargar_historico(archivo)
    r = np.log(h["trm"]).diff()
    r.index = h["fecha"]
    r = r.dropna()
    x = r[r.index > pd.Timestamp(fecha_desde)].values
    nu, loc, esc = stats.t.fit(x)
    return nu


def simular(mu_q, sig_q, seed=SEED, dist="normal", nu=None):
    """Trayectorias de la TRM.
    dist="normal": innovaciones normales (versión original, mismos sorteos).
    dist="t": innovaciones t de Student con nu grados de libertad, reescaladas para
    tener la misma varianza que la normal; en el año 1 se conserva el rango de cada
    escenario como intervalo de 80 % (usando el cuantil de la t), así que solo
    cambian las colas."""
    rng = np.random.default_rng(seed)
    probs = np.array([e[1] for e in ESCENARIOS])
    idx = rng.choice(len(ESCENARIOS), size=N_SIM, p=probs)
    meds = np.array([mediana_y_sigma(e[2], e[3])[0] for e in ESCENARIOS])
    sigs = np.array([mediana_y_sigma(e[2], e[3])[1] for e in ESCENARIOS])

    if dist == "normal":
        choque = lambda size: rng.standard_normal(size)
        choque_fin1 = lambda size: rng.standard_normal(size)
    elif dist == "t":
        from scipy import stats
        escala = np.sqrt((nu - 2) / nu)                 # varianza unitaria
        choque = lambda size: rng.standard_t(nu, size) * escala
        z80_t = stats.t.ppf(0.90, nu)
        sigs = sigs * Z80 / z80_t                        # mismo intervalo de 80 %
        choque_fin1 = lambda size: rng.standard_t(nu, size)
    else:
        raise ValueError("dist debe ser 'normal' o 't'")

    logS = np.empty((N_SIM, N_Q + 1))
    logS[:, 0] = np.log(S0)
    # Fin del año 1
    end1 = np.log(meds[idx]) + sigs[idx] * choque_fin1(N_SIM)
    # Puente browniano en los trimestres 1..3
    sq1 = VOL_ANIO1 / 2.0
    W = np.cumsum(sq1 * choque((N_SIM, 4)), axis=1)
    t = np.arange(1, 5)
    bridge = W - (t / 4.0) * W[:, [3]]
    for k in range(1, 5):
        logS[:, k] = logS[:, 0] + (end1 - logS[:, 0]) * k / 4 + bridge[:, k - 1]
    # Años 2-10: parámetros históricos
    for k in range(5, N_Q + 1):
        logS[:, k] = logS[:, k - 1] + mu_q + sig_q * choque(N_SIM)
    return np.exp(logS), idx


def tabla_credito():
    monto_usd = VALOR_MAQUINARIA_COP * (1 - APORTE) / S0
    i = TASA_NOMINAL / PAGOS_ANIO
    n = N_Q
    cuota = monto_usd * i / (1 - (1 + i) ** -n)
    saldo = monto_usd
    filas = []
    for k in range(1, n + 1):
        interes = saldo * i
        capital = cuota - interes
        saldo -= capital
        filas.append((k, interes, capital, saldo))
    df = pd.DataFrame(filas, columns=["trimestre", "interes_usd", "capital_usd", "saldo_usd"])
    return monto_usd, cuota, df


def tir_anual(flujos):
    """TIR trimestral por bisección (flujo inicial positivo = desembolso recibido),
    expresada como tasa efectiva anual."""
    lo, hi = -0.5, 1.0
    t = np.arange(len(flujos))
    for _ in range(200):
        mid = (lo + hi) / 2
        vpn = np.sum(flujos / (1 + mid) ** t)
        if vpn > 0:      # la tasa es demasiado alta
            hi = mid
        else:
            lo = mid
    return (1 + mid) ** 4 - 1


def costos_cop(paths, cuota, amort, monto_usd):
    """Costo en pesos por trayectoria: total pagado, efecto cambiario y costo efectivo anual."""
    S = paths[:, 1:]
    total = cuota * S.sum(axis=1)
    base_spot = cuota * N_Q * S0
    efecto_cambiario = total - base_spot
    efecto_interes = (amort["interes_usd"].values * (S - S0)).sum(axis=1)
    efecto_capital = (amort["capital_usd"].values * (S - S0)).sum(axis=1)
    return total, efecto_cambiario, efecto_interes, efecto_capital


def _num(x, dec=2):
    """Formato colombiano: 3.174,67"""
    return f"{x:,.{dec}f}".replace(",", "X").replace(".", ",").replace("X", ".")


def _mostrar(tabla, titulo):
    """Muestra una tabla: en Colab/Jupyter como tabla HTML; en consola como texto."""
    print("\n" + titulo)
    try:
        from IPython import get_ipython
        from IPython.display import display
        if get_ipython() is not None:
            display(tabla)
            return
    except ImportError:
        pass
    with pd.option_context("display.width", 250, "display.max_columns", 30, "display.max_rows", 100):
        print(tabla.to_string(index=False))


def tablas_40_trimestres(df, paths, cuota, amort):
    """Valores proyectados de los 40 trimestres: TRM y cuota en pesos."""
    # --- TRM ---
    t_trm = df.copy()
    t_trm.to_csv("tabla_trm_40_trimestres.csv", index=False)
    cols = {"trimestre": "Trim.", "fecha": "Fecha", "spot_constante": "TRM spot",
            "escenario_base": "Base", "escenario_peso_fuerte": "Peso fuerte",
            "escenario_peso_debil": "Peso débil", "sim_media": "Sim. media", "sim_p05": "Sim. P5",
            "sim_p25": "Sim. P25", "sim_p50": "Sim. P50", "sim_p75": "Sim. P75", "sim_p95": "Sim. P95"}
    f_trm = t_trm.rename(columns=cols)
    for c in list(cols.values())[2:]:
        f_trm[c] = f_trm[c].map(_num)
    _mostrar(f_trm, "TRM proyectada por trimestre (COP por USD)")

    # --- cuotas en pesos ---
    S = paths[:, 1:]
    cuotas = cuota * S
    pc = np.percentile(cuotas, [5, 50, 95], axis=0)
    t_c = pd.DataFrame({
        "trimestre": np.arange(1, N_Q + 1),
        "fecha": df["fecha"].values[1:],
        "cuota_usd": cuota,
        "interes_usd": amort["interes_usd"].values,
        "capital_usd": amort["capital_usd"].values,
        "cuota_cop_spot": cuota * S0,
        "cuota_cop_base": cuota * df["escenario_base"].values[1:],
        "cuota_cop_peso_fuerte": cuota * df["escenario_peso_fuerte"].values[1:],
        "cuota_cop_peso_debil": cuota * df["escenario_peso_debil"].values[1:],
        "cuota_cop_sim_media": cuotas.mean(axis=0),
        "cuota_cop_sim_p05": pc[0], "cuota_cop_sim_p50": pc[1], "cuota_cop_sim_p95": pc[2],
    })
    t_c.to_csv("tabla_cuotas_cop_40_trimestres.csv", index=False)
    f_c = pd.DataFrame({
        "Trim.": t_c.trimestre, "Fecha": t_c.fecha,
        "Cuota USD": t_c.cuota_usd.map(_num), "Interés USD": t_c.interes_usd.map(_num),
        "Capital USD": t_c.capital_usd.map(_num),
    })
    for c, nombre in [("cuota_cop_spot", "Spot"), ("cuota_cop_base", "Base"), ("cuota_cop_peso_fuerte", "Peso fuerte"),
                      ("cuota_cop_peso_debil", "Peso débil"), ("cuota_cop_sim_media", "Sim. media"),
                      ("cuota_cop_sim_p05", "Sim. P5"), ("cuota_cop_sim_p50", "Sim. P50"), ("cuota_cop_sim_p95", "Sim. P95")]:
        f_c[nombre] = (t_c[c] / 1e6).map(_num)
    total = {"Trim.": "Total", "Fecha": "", "Cuota USD": _num(t_c.cuota_usd.sum()),
             "Interés USD": _num(t_c.interes_usd.sum()), "Capital USD": _num(t_c.capital_usd.sum())}
    for c, nombre in [("cuota_cop_spot", "Spot"), ("cuota_cop_base", "Base"), ("cuota_cop_peso_fuerte", "Peso fuerte"),
                      ("cuota_cop_peso_debil", "Peso débil"), ("cuota_cop_sim_media", "Sim. media")]:
        total[nombre] = _num(t_c[c].sum() / 1e6)
    for nombre in ["Sim. P5", "Sim. P50", "Sim. P95"]:
        total[nombre] = "–"          # los percentiles por trimestre no se suman
    f_c = pd.concat([f_c.astype(str), pd.DataFrame([total])], ignore_index=True)
    _mostrar(f_c, "Cuota de cada trimestre: en USD y en millones de pesos")
    return t_c


def graficas(df, paths, cuota, amort, paths_por_ventana, total):
    import matplotlib.pyplot as plt
    from matplotlib.ticker import FuncFormatter
    fmt = FuncFormatter(lambda v, _: _num(v, 0))
    fmt1 = FuncFormatter(lambda v, _: _num(v, 1))
    x = df["trimestre"].values
    fig = plt.figure(figsize=(14, 15))
    gs = fig.add_gridspec(3, 2, hspace=0.35, wspace=0.22)

    # 1) TRM
    ax = fig.add_subplot(gs[0, 0])
    rng = np.random.default_rng(7)
    for j in rng.choice(paths.shape[0], 25, replace=False):
        ax.plot(x, paths[j], color="#9db9d8", lw=0.5, alpha=0.6)
    ax.fill_between(x, df.sim_p05, df.sim_p95, color="#9db9d8", alpha=0.35, lw=0, label="Simulación P5–P95")
    ax.fill_between(x, df.sim_p25, df.sim_p75, color="#5b8cc0", alpha=0.45, lw=0, label="Simulación P25–P75")
    ax.plot(x, df.sim_p50, color="#1f3d63", lw=2, label="Simulación mediana")
    ax.plot(x, df.sim_media, color="#1f3d63", lw=1.2, ls=":", label="Simulación media")
    ax.plot(x, df.spot_constante, color="#555", ls="--", lw=1.2, label="TRM spot constante")
    ax.plot(x, df.escenario_peso_fuerte, color="#2e8b57", lw=1.3, label="Escenario peso fuerte")
    ax.plot(x, df.escenario_base, color="#e67e22", lw=1.3, label="Escenario base")
    ax.plot(x, df.escenario_peso_debil, color="#c0392b", lw=1.3, label="Escenario peso débil")
    ax.set_title("TRM proyectada (COP por USD)"); ax.set_xlabel("Trimestre"); ax.yaxis.set_major_formatter(fmt)
    ax.legend(fontsize=8, frameon=False, loc="upper left")

    # 2) Cuota en pesos
    ax = fig.add_subplot(gs[0, 1])
    k = x[1:]
    c = cuota * paths[:, 1:] / 1e6
    p5, p25, p50, p75, p95 = np.percentile(c, [5, 25, 50, 75, 95], axis=0)
    ax.fill_between(k, p5, p95, color="#9db9d8", alpha=0.35, lw=0, label="Simulación P5–P95")
    ax.fill_between(k, p25, p75, color="#5b8cc0", alpha=0.45, lw=0, label="Simulación P25–P75")
    ax.plot(k, p50, color="#1f3d63", lw=2, label="Simulación mediana")
    ax.plot(k, cuota * df.spot_constante.values[1:] / 1e6, color="#555", ls="--", lw=1.2, label="Con TRM spot")
    ax.plot(k, cuota * df.escenario_base.values[1:] / 1e6, color="#e67e22", lw=1.3, label="Escenario base")
    ax.set_title(f"Cuota trimestral en pesos (USD {_num(cuota)} constantes)"); ax.set_xlabel("Trimestre")
    ax.set_ylabel("Millones de COP"); ax.yaxis.set_major_formatter(fmt1); ax.legend(fontsize=8, frameon=False, loc="upper left")

    # 3) Ventanas históricas
    ax = fig.add_subplot(gs[1, 0])
    colores = ["#c0392b", "#e67e22", "#8e44ad", "#16a085", "#1f3d63", "#7f8c8d"]
    for (nombre, (pth, _, prm)), col in zip(paths_por_ventana.items(), colores):
        lw = 2.4 if nombre == VENTANA_ELEGIDA else 1.2
        ax.plot(x, np.percentile(pth, 50, axis=0), color=col, lw=lw,
                label=f"{nombre}: deriva {_num(prm['deriva_anual']*100, 1)} %, vol. {_num(prm['volatilidad_anual']*100, 1)} %")
    ax.axhline(S0, color="#555", ls="--", lw=1)
    ax.set_title("TRM mediana según la ventana histórica usada"); ax.set_xlabel("Trimestre")
    ax.yaxis.set_major_formatter(fmt); ax.legend(fontsize=7.5, frameon=False, loc="upper left")

    # 4) Histograma del costo total
    ax = fig.add_subplot(gs[1, 1])
    ax.hist(total / 1e6, bins=80, color="#5b8cc0", alpha=0.8)
    spot_total = cuota * N_Q * S0 / 1e6
    for v, lab, col, ls in [(spot_total, "TRM spot", "#555", "--"), (total.mean() / 1e6, "Media", "#1f3d63", "-"),
                            (np.percentile(total, 5) / 1e6, "P5", "#2e8b57", ":"), (np.percentile(total, 95) / 1e6, "P95", "#c0392b", ":")]:
        ax.axvline(v, color=col, ls=ls, lw=1.5, label=f"{lab}: {_num(v, 1)}")
    ax.set_title("Costo total del crédito en pesos (20.000 simulaciones)"); ax.set_xlabel("Millones de COP")
    ax.set_yticks([]); ax.xaxis.set_major_formatter(fmt); ax.legend(fontsize=8, frameon=False)

    # 5) Intereses vs tasa de cambio (escenario base)
    ax = fig.add_subplot(gs[2, :])
    Sb = df.escenario_base.values[1:]
    i_spot = amort["interes_usd"].values * S0 / 1e6
    c_spot = amort["capital_usd"].values * S0 / 1e6
    ef_tc = cuota * (Sb - S0) / 1e6
    ax.bar(k, c_spot, color="#5b8cc0", label="Capital valorado a TRM spot")
    ax.bar(k, i_spot, bottom=c_spot, color="#e67e22", label="Intereses valorados a TRM spot")
    ax.bar(k, ef_tc, bottom=c_spot + i_spot, color="#c0392b", label="Efecto de la tasa de cambio")
    ax.set_title("Escenario base: cuota en pesos separada entre capital, intereses y efecto cambiario")
    ax.set_xlabel("Trimestre"); ax.set_ylabel("Millones de COP"); ax.yaxis.set_major_formatter(fmt1)
    ax.legend(fontsize=8.5, frameon=False, loc="upper left", ncol=3)

    for a in fig.axes:
        a.grid(alpha=0.25); a.spines[["top", "right"]].set_visible(False)
    fig.savefig("graficas_simulacion_trm.png", dpi=150, bbox_inches="tight")
    plt.show()


if __name__ == "__main__":
    monto_usd, cuota, amort = tabla_credito()
    print(f"Monto USD {monto_usd:,.2f} | cuota trimestral USD {cuota:,.2f}\n")

    # 1) Parámetros y resultados para cada ventana histórica
    filas = []
    paths_por_ventana = {}
    for nombre, desde in VENTANAS.items():
        p = estimar_parametros(desde)
        paths, idx = simular(p["mu_q"], p["sig_q"])
        paths_por_ventana[nombre] = (paths, idx, p)
        total, ef_tc, _, _ = costos_cop(paths, cuota, amort, monto_usd)
        filas.append({
            "ventana": nombre, "n_trimestres": p["n_trimestres"],
            "deriva_anual": p["deriva_anual"], "volatilidad_anual": p["volatilidad_anual"],
            "error_est_deriva": p["error_est_deriva_anual"],
            "TRM_p50_anio5": np.percentile(paths[:, 20], 50),
            "TRM_p05_anio10": np.percentile(paths[:, 40], 5),
            "TRM_p50_anio10": np.percentile(paths[:, 40], 50),
            "TRM_p95_anio10": np.percentile(paths[:, 40], 95),
            "costo_total_cop_media": total.mean(),
            "costo_total_cop_p95": np.percentile(total, 95),
        })
    tab = pd.DataFrame(filas)
    tab.to_csv("parametros_por_ventana.csv", index=False)
    with pd.option_context("display.width", 200, "display.float_format", "{:,.4f}".format):
        print(tab.to_string(index=False))

    # 2) Ventana elegida: trayectorias, escenarios y cuotas en pesos
    paths, idx, p = paths_por_ventana[VENTANA_ELEGIDA]
    mu_q = p["mu_q"]
    pct = np.percentile(paths, [5, 25, 50, 75, 95], axis=0)
    fechas = pd.date_range(FECHA_INICIO, periods=N_Q + 1, freq=pd.DateOffset(months=3))
    df = pd.DataFrame({
        "trimestre": np.arange(N_Q + 1),
        "fecha": fechas.date,
        "spot_constante": S0,
        "escenario_base": trayectoria_escenario_mediana(mediana_y_sigma(*ESCENARIOS[0][2:])[0], mu_q),
        "escenario_peso_fuerte": trayectoria_escenario_mediana(mediana_y_sigma(*ESCENARIOS[1][2:])[0], mu_q),
        "escenario_peso_debil": trayectoria_escenario_mediana(mediana_y_sigma(*ESCENARIOS[2][2:])[0], mu_q),
        "sim_media": paths.mean(axis=0),
        "sim_p05": pct[0], "sim_p25": pct[1], "sim_p50": pct[2], "sim_p75": pct[3], "sim_p95": pct[4],
    })
    df.to_csv("trm_trayectorias.csv", index=False)
    np.save("trm_paths.npy", paths.astype("float32"))

    cuotas_cop = cuota * paths[:, 1:]
    pc = np.percentile(cuotas_cop, [5, 50, 95], axis=0)
    pd.DataFrame({
        "trimestre": np.arange(1, N_Q + 1),
        "cuota_cop_spot": cuota * S0,
        "cuota_cop_media": cuotas_cop.mean(axis=0),
        "cuota_cop_p05": pc[0], "cuota_cop_p50": pc[1], "cuota_cop_p95": pc[2],
    }).to_csv("cuotas_cop_simulacion.csv", index=False)

    total, ef_tc, ef_int, ef_cap = costos_cop(paths, cuota, amort, monto_usd)
    monto_cop = VALOR_MAQUINARIA_COP * (1 - APORTE)
    tir = np.array([tir_anual(np.concatenate([[monto_cop], -cuota * paths[j, 1:]]))
                    for j in range(0, N_SIM, 20)])   # muestra de 1.000 trayectorias
    print(f"\nVentana elegida: {VENTANA_ELEGIDA} | deriva {p['deriva_anual']:.4f} | vol {p['volatilidad_anual']:.4f}")
    print(df.iloc[[0, 1, 2, 4, 8, 12, 20, 40]].round(1).to_string(index=False))
    print("\nProporción de sorteos por escenario:",
          {ESCENARIOS[i][0]: round(float((idx == i).mean()), 3) for i in range(3)})
    print("Costo total en COP: media %.0f | P5 %.0f | P50 %.0f | P95 %.0f" % (
        total.mean(), *np.percentile(total, [5, 50, 95])))
    print("Efecto cambiario: media %.0f (sobre intereses %.0f, sobre capital %.0f)" % (
        ef_tc.mean(), ef_int.mean(), ef_cap.mean()))
    print("Probabilidad de pagar más que con la TRM spot: %.3f" % (ef_tc > 0).mean())
    print("Costo efectivo anual en COP (muestra): media %.4f | P5 %.4f | P50 %.4f | P95 %.4f" % (
        tir.mean(), *np.percentile(tir, [5, 50, 95])))

    # 3) Valores proyectados de los 40 trimestres y gráficas
    if MOSTRAR_TABLAS:
        tablas_40_trimestres(df, paths, cuota, amort)
    if MOSTRAR_GRAFICAS:
        graficas(df, paths, cuota, amort, paths_por_ventana, total)
```

```python
"""
Parte B: cobertura con cuatro forwards anuales consecutivos (años 7 a 10).

Supuestos (ver hoja «B Supuestos cobertura» del Excel):
  - Al terminar el año 6 (trimestre 24) quedan 16 cuotas de USD 5.133,11 (TRM inicial 3.157,94).
  - Se cubre el 75 % de esa exposición, repartido según los pagos de cada año:
    cuatro forwards de 0,75 × 4 cuotas = USD 15.399,33 cada uno.
  - Forwards consecutivos de un año, liquidados por compensación (NDF):
      forward 1: se pacta en el trimestre 24 y vence en el 28 (año 7)
      forward 2: se pacta en el 28 y vence en el 32 (año 8)
      forward 3: se pacta en el 32 y vence en el 36 (año 9)
      forward 4: se pacta en el 36 y vence en el 40 (año 10)
    Precio pactado F = S(pacto) × (1 + i_COP) / (1 + i_USD)   (paridad cubierta, 1 año)
    Liquidación al vencimiento (COP) = N × (S(vencimiento) − F)
  - Las cuotas trimestrales se compran siempre al spot; el NDF compensa la diferencia.
  - Variante de comparación («strip»): los cuatro precios se fijan en el trimestre 24
    con plazos de 1, 2, 3 y 4 años: F_k = S(24) × [(1 + i_COP)/(1 + i_USD)]^k.

Las trayectorias de la TRM vienen de simulacion_trm.py (ventana de flotación libre)
con innovaciones normales y con innovaciones t de Student.

Salidas: cobertura_resultados.csv, cobertura_por_forward.csv, sensibilidad_tasas.csv,
trayectorias_representativas.csv y graficas_cobertura_forward.png.

Uso: python3 cobertura_forward.py   (en la misma carpeta que simulacion_trm.py), o en
Colab/Jupyter: ejecutar primero una celda con el código de simulacion_trm.py y después
otra con este código.
"""
import numpy as np
import pandas as pd
try:
    import simulacion_trm as sim
except ModuleNotFoundError:
    # En Colab/Jupyter: si el código de simulacion_trm.py se ejecutó en una celda
    # anterior, sus funciones ya están en el notebook y se usan desde ahí.
    import __main__ as sim
    if not hasattr(sim, "simular"):
        raise ModuleNotFoundError(
            "Ejecute primero el código de simulacion_trm.py en una celda anterior "
            "o suba el archivo simulacion_trm.py a la carpeta de trabajo.")

# ----------------------------- PARÁMETROS ---------------------------------
# Spot de valoración: TRM del 27/10/2026 (Banco de la República). Es la misma TRM con la que
# empieza la parte A, para que las dos partes usen la misma fecha.
SPOT_MERCADO = sim.S0                       # 3.157,94
# Puntos forward a 1 año de la curva de Cbonds (precio medio, cierre de Londres, 27/10/2026):
# forward 1 año 3.386,30 − spot S/N 3.132,07. Se suman a la TRM para obtener el forward observado.
PUNTOS_1Y = 3386.30 - 3132.07               # 254,23 pesos
FWD_1Y_MERCADO = SPOT_MERCADO + PUNTOS_1Y   # 3.412,17
# Todas las tasas en efectiva anual (EA) para que la paridad sea consistente
TESORO_1A = 0.0447              # Tesoro de EE. UU. a 1 año, rendimiento de bono (Fed H.15, 26/10/2026)
I_USD = (1 + TESORO_1A / 2) ** 2 - 1        # = 4,52 % EA
IBR_12M_NOMINAL = 0.12478       # IBR a 12 meses del 27/10/2026, nominal base 360 (Banco de la República, serie 241 de suameca)
DIAS_12M = 365                  # días reales del 27/10/2026 al 27/10/2027
# Conversión a efectiva anual según el manual de la IBR: EA = (1 + i·n/360)^(365/n) − 1; con n = 365: i × 365/360
I_COP_IBR = (1 + IBR_12M_NOMINAL * DIAS_12M / 360) ** (365 / DIAS_12M) - 1   # = 12,651 % EA
# Tasa en pesos implícita en el forward de mercado (base del modelo: es el precio que
# pagaría una empresa). La prima forward resultante es la del mercado: F/S − 1.
I_COP = (FWD_1Y_MERCADO / SPOT_MERCADO) * (1 + I_USD) - 1

COBERTURA = 0.75
TRIM_INICIO = 24                # fin del año 6
PACTOS = [24, 28, 32, 36]
VENCIMIENTOS = [28, 32, 36, 40]
UMBRAL_EXTREMO = 0.20           # depreciación anual > 20 % = pérdida cambiaria extrema
TOLERANCIA_PPTO = 0.10          # «fracaso»: costo cubierto > presupuesto + 10 %
TASA_DESCUENTO_EA = 0.13        # para el VPN en pesos (TES a 10 años > 13 %)
NU_STRESS = 4                   # t de Student de estrés (colas mucho más gruesas)
MOSTRAR_GRAFICAS = True         # dibuja las gráficas (en Colab/Jupyter aparecen debajo de la celda)
# --------------------------------------------------------------------------


def prima(i_cop, i_usd):
    return (1 + i_cop) / (1 + i_usd) - 1


def evaluar(paths, cuota, i_cop=I_COP, i_usd=I_USD, variante="rodante"):
    """Resultados por trayectoria de la estrategia de cobertura."""
    n = paths.shape[0]
    p = prima(i_cop, i_usd)
    N = COBERTURA * 4 * cuota                                  # nominal por forward
    S = paths
    F = np.empty((n, 4)); Sv = np.empty((n, 4)); Sp = np.empty((n, 4))
    U = np.empty((n, 4))
    for k, (c, m) in enumerate(zip(PACTOS, VENCIMIENTOS)):
        Sp[:, k] = S[:, c]
        Sv[:, k] = S[:, m]
        if variante == "rodante":
            F[:, k] = S[:, c] * (1 + p)
        else:                                                    # strip fijado en t=24
            F[:, k] = S[:, TRIM_INICIO] * (1 + p) ** (k + 1)
        U[:, k] = cuota * S[:, c + 1:m + 1].sum(axis=1)          # costo del año sin cobertura
    liq = N * (Sv - F)                                           # liquidación NDF (COP)
    H = U - liq                                                  # costo del año con cobertura
    B = 4 * cuota * F                                            # presupuesto al forward
    # flujos trimestrales del crédito (pagos positivos = salidas)
    pagos_sin = cuota * S[:, 1:]
    pagos_con = pagos_sin.copy()
    for k, m in enumerate(VENCIMIENTOS):
        pagos_con[:, m - 1] -= liq[:, k]
    return dict(F=F, Sv=Sv, Sp=Sp, U=U, H=H, B=B, liq=liq, N=N, p=p,
                pagos_sin=pagos_sin, pagos_con=pagos_con)


def tir_ea(desembolso, pagos):
    """TIR trimestral (vectorizada por bisección) expresada como tasa efectiva anual."""
    n = pagos.shape[0]
    lo = np.full(n, -0.2); hi = np.full(n, 0.5)
    t = np.arange(1, pagos.shape[1] + 1)
    for _ in range(80):
        mid = (lo + hi) / 2
        vpn = desembolso - (pagos / (1 + mid[:, None]) ** t).sum(axis=1)
        hi = np.where(vpn > 0, mid, hi)
        lo = np.where(vpn > 0, lo, mid)
    return (1 + mid) ** 4 - 1


def vpn(pagos, tasa_ea=TASA_DESCUENTO_EA):
    iq = (1 + tasa_ea) ** 0.25 - 1
    t = np.arange(1, pagos.shape[1] + 1)
    return (pagos / (1 + iq) ** t).sum(axis=1)


def cvar(x, q=0.95):
    v = np.quantile(x, q)
    return x[x >= v].mean()


def resumen(paths, cuota, etiqueta, desembolso):
    r = evaluar(paths, cuota)
    rs = evaluar(paths, cuota, variante="strip")
    filas = {}
    dep = r["Sv"] / r["Sp"] - 1                                  # depreciación del año del forward
    # --- B.2: proporciones por forward (agrupando los cuatro años) ---
    prot = (r["Sv"] > r["F"]).mean()
    opor = (r["Sv"] < r["F"]).mean()
    ext = dep > UMBRAL_EXTREMO
    filas["Proporción con protección (spot al vencimiento > forward)"] = prot
    filas["Proporción con costo de oportunidad (spot < forward)"] = opor
    filas["Proporción con pérdida cambiaria extrema (depreciación anual > 20 %)"] = ext.mean()
    filas["Proporción con depreciación anual > 30 %"] = (dep > 0.30).mean()
    filas["Proporción con apreciación anual > 20 % (costo de oportunidad extremo)"] = (dep < -0.20).mean()
    filas["En años extremos: sobrecosto sin cobertura vs presupuesto"] = (r["U"][ext] / r["B"][ext] - 1).mean() if ext.any() else np.nan
    filas["En años extremos: sobrecosto con cobertura vs presupuesto"] = (r["H"][ext] / r["B"][ext] - 1).mean() if ext.any() else np.nan
    base_pacto = 4 * cuota * r["Sp"]
    filas["En años extremos: sobrecosto sin cobertura vs spot del pacto"] = (r["U"][ext] / base_pacto[ext] - 1).mean()
    filas["En años extremos: sobrecosto con cobertura vs spot del pacto"] = (r["H"][ext] / base_pacto[ext] - 1).mean()
    filas["En años extremos: pérdida sin cobertura vs spot del pacto (COP)"] = (r["U"][ext] - base_pacto[ext]).mean()
    filas["En años extremos: pérdida con cobertura vs spot del pacto (COP)"] = (r["H"][ext] - base_pacto[ext]).mean()
    perd_der = r["liq"] < 0
    fr_sin = r["U"] > r["B"] * (1 + TOLERANCIA_PPTO)
    fr_con = r["H"] > r["B"] * (1 + TOLERANCIA_PPTO)
    filas["Pérdida del derivado (liquidación negativa)"] = perd_der.mean()
    filas["Pérdida del derivado y cobertura exitosa"] = (perd_der & ~fr_con).mean()
    filas["Ganancia del derivado y cobertura exitosa"] = (~perd_der & ~fr_con).mean()
    filas["Fracaso de la cobertura (costo cubierto > presupuesto + 10 %)"] = fr_con.mean()
    filas["Sin cobertura: costo > presupuesto + 10 %"] = fr_sin.mean()
    # --- B.3: costo total del crédito ---
    for nombre, pagos in [("sin cobertura", r["pagos_sin"]), ("con cobertura rodante", r["pagos_con"]),
                          ("con strip fijado en año 6", rs["pagos_con"])]:
        tot = pagos.sum(axis=1)
        a710 = pagos[:, 24:].sum(axis=1)
        filas[f"Costo total COP – {nombre}: media"] = tot.mean()
        filas[f"Costo total COP – {nombre}: desv. estándar"] = tot.std()
        filas[f"Costo total COP – {nombre}: P5"] = np.quantile(tot, 0.05)
        filas[f"Costo total COP – {nombre}: P95"] = np.quantile(tot, 0.95)
        filas[f"Costo total COP – {nombre}: P99"] = np.quantile(tot, 0.99)
        filas[f"Costo total COP – {nombre}: CVaR 95 %"] = cvar(tot)
        filas[f"Costo años 7-10 – {nombre}: media"] = a710.mean()
        filas[f"Costo años 7-10 – {nombre}: desv. estándar"] = a710.std()
        filas[f"Costo años 7-10 – {nombre}: P95"] = np.quantile(a710, 0.95)
        filas[f"Costo años 7-10 – {nombre}: CVaR 95 %"] = cvar(a710)
        filas[f"VPN pagos al 13 % EA – {nombre}: media"] = vpn(pagos).mean()
        tir = tir_ea(desembolso, pagos)
        filas[f"Costo efectivo anual – {nombre}: media"] = tir.mean()
        filas[f"Costo efectivo anual – {nombre}: P95"] = np.quantile(tir, 0.95)
    # riesgo visto desde el fin del año 6 (cuando se toma la decisión): costo de los años
    # 7-10 relativo al costo de las 16 cuotas a la TRM del trimestre 24
    base24 = 16 * cuota * paths[:, TRIM_INICIO]
    for nombre, pagos in [("sin cobertura", r["pagos_sin"]), ("con cobertura rodante", r["pagos_con"]),
                          ("con strip fijado en año 6", rs["pagos_con"])]:
        rel = pagos[:, 24:].sum(axis=1) / base24 - 1
        filas[f"Desde el año 6 – {nombre}: sobrecosto medio vs TRM del año 6"] = rel.mean()
        filas[f"Desde el año 6 – {nombre}: desv. estándar"] = rel.std()
        filas[f"Desde el año 6 – {nombre}: P95"] = np.quantile(rel, 0.95)
        filas[f"Desde el año 6 – {nombre}: P99"] = np.quantile(rel, 0.99)
    # desviación anual frente al presupuesto (estabilidad presupuestal, por año del forward)
    dev_sin = r["U"] / r["B"] - 1
    dev_con = r["H"] / r["B"] - 1
    filas["Desv. estándar de la desviación anual vs presupuesto – sin cobertura"] = dev_sin.std()
    filas["Desv. estándar de la desviación anual vs presupuesto – con cobertura"] = dev_con.std()
    filas["Resultado acumulado de los 4 forwards: media (COP)"] = r["liq"].sum(axis=1).mean()
    filas["Resultado acumulado de los 4 forwards: P5 (COP)"] = np.quantile(r["liq"].sum(axis=1), 0.05)
    filas["Resultado acumulado de los 4 forwards: P95 (COP)"] = np.quantile(r["liq"].sum(axis=1), 0.95)
    filas["Prob. de que la cobertura rodante cueste más que no cubrir"] = (r["liq"].sum(axis=1) < 0).mean()
    filas["Strip: resultado acumulado de los 4 forwards, media (COP)"] = rs["liq"].sum(axis=1).mean()
    filas["Strip: prob. de que cueste más que no cubrir"] = (rs["liq"].sum(axis=1) < 0).mean()
    return pd.Series(filas, name=etiqueta), r


def por_forward(r):
    filas = []
    for k in range(4):
        filas.append({
            "forward": k + 1, "año": 7 + k,
            "F_mediana": np.median(r["F"][:, k]), "S_venc_mediana": np.median(r["Sv"][:, k]),
            "prob_proteccion": (r["Sv"][:, k] > r["F"][:, k]).mean(),
            "prob_costo_oportunidad": (r["Sv"][:, k] < r["F"][:, k]).mean(),
            "prob_extrema": (r["Sv"][:, k] / r["Sp"][:, k] - 1 > UMBRAL_EXTREMO).mean(),
            "liq_media": r["liq"][:, k].mean(), "liq_p5": np.quantile(r["liq"][:, k], 0.05),
            "liq_p95": np.quantile(r["liq"][:, k], 0.95),
            "costo_sin_media": r["U"][:, k].mean(), "costo_con_media": r["H"][:, k].mean(),
            "costo_sin_sd": r["U"][:, k].std(), "costo_con_sd": r["H"][:, k].std(),
        })
    return pd.DataFrame(filas)


def _num(x, dec=2):
    """Formato colombiano: 3.174,67"""
    return f"{x:,.{dec}f}".replace(",", "X").replace(".", ",").replace("X", ".")


def graficas(paths_n, cuota, r_n, r_t, r_t4, rs_n, sens, rep, nu):
    """Gráficas de la parte B."""
    import matplotlib.pyplot as plt
    from matplotlib.ticker import FuncFormatter
    fmt0 = FuncFormatter(lambda v, _: _num(v, 0))
    fmt1 = FuncFormatter(lambda v, _: _num(v, 1))
    pct = FuncFormatter(lambda v, _: _num(v, 0) + " %")
    AZ, GR, VE, RO, NA = "#1f5f9e", "#8a8a8a", "#2e8b57", "#c0392b", "#e67e22"
    fig = plt.figure(figsize=(18, 16))
    gs = fig.add_gridspec(3, 3, hspace=0.42, wspace=0.25, top=0.92)

    # (a) Forward pactado vs TRM al vencimiento
    ax = fig.add_subplot(gs[0, 0])
    rng = np.random.default_rng(3)
    j = rng.choice(r_n["F"].size, 4000, replace=False)
    F = r_n["F"].ravel()[j]; Sv = r_n["Sv"].ravel()[j]
    prot = Sv > F
    ax.scatter(F[prot], Sv[prot], s=4, alpha=0.4, color=VE, label=f"Protección ({(r_n['Sv'] > r_n['F']).mean()*100:.1f} %)".replace(".", ","))
    ax.scatter(F[~prot], Sv[~prot], s=4, alpha=0.4, color=RO, label=f"Costo de oportunidad ({(r_n['Sv'] < r_n['F']).mean()*100:.1f} %)".replace(".", ","))
    lim = [min(F.min(), Sv.min()), np.percentile(np.r_[F, Sv], 99.5)]
    ax.plot(lim, lim, color="#333", lw=1, ls="--", label="TRM al vencimiento = forward")
    ax.set_xlim(lim); ax.set_ylim(lim)
    ax.set_title("(a) Forward pactado vs TRM al vencimiento\n(4.000 contratos simulados, normal)", fontsize=11)
    ax.set_xlabel("Forward pactado (COP/USD)"); ax.set_ylabel("TRM al vencimiento (COP/USD)")
    ax.xaxis.set_major_formatter(fmt0); ax.yaxis.set_major_formatter(fmt0); ax.legend(fontsize=8, frameon=False, loc="upper left")

    # (b) TRM al vencimiento / forward − 1, normal vs t
    ax = fig.add_subplot(gs[0, 1])
    bins = np.linspace(-45, 60, 85)
    for r, lab, col in [(r_n, "Normal", AZ), (r_t, f"t (ν = {_num(nu, 1)})", NA), (r_t4, "t de estrés (ν = 4)", RO)]:
        ax.hist((r["Sv"] / r["F"] - 1).ravel() * 100, bins=bins, histtype="step", lw=1.5, density=True, color=col, label=lab)
    ax.axvline(0, color="#333", lw=1, ls="--")
    ax.set_ylim(0, ax.get_ylim()[1] * 1.3)
    ax.text(1.5, ax.get_ylim()[1] * 0.93, "protección →", fontsize=8, color=VE)
    ax.text(-1.5, ax.get_ylim()[1] * 0.93, "← costo de oportunidad", fontsize=8, color=RO, ha="right")
    ax.set_title("(b) Diferencia entre la TRM al vencimiento y el forward", fontsize=11)
    ax.set_xlabel("TRM al vencimiento / forward − 1"); ax.xaxis.set_major_formatter(pct); ax.set_yticks([])
    ax.set_ylim(0, ax.get_ylim()[1] * 1.3)
    ax.legend(fontsize=8, frameon=False, loc="center right")

    # (c) Resultado acumulado de los 4 forwards
    ax = fig.add_subplot(gs[0, 2])
    bins = np.linspace(-80, 80, 81)
    for r, lab, col in [(r_n, "Normal", AZ), (r_t, f"t (ν = {_num(nu, 1)})", NA)]:
        liq = r["liq"].sum(axis=1) / 1e6
        ax.hist(liq, bins=bins, histtype="step", lw=1.5, density=True, color=col,
                label=f"{lab}: media {_num(liq.mean(), 1)} millones")
    ax.axvline(0, color="#333", lw=1, ls="--")
    ax.set_title("(c) Resultado acumulado de los cuatro forwards", fontsize=11)
    ax.set_xlabel("Millones de COP (positivo = la cobertura ahorra)"); ax.set_yticks([])
    ax.set_ylim(0, ax.get_ylim()[1] * 1.25)
    ax.xaxis.set_major_formatter(fmt0); ax.legend(fontsize=8, frameon=False, loc="upper right")

    # (d) Costo años 7-10 visto desde el año 6
    ax = fig.add_subplot(gs[1, 0])
    base24 = 16 * cuota * paths_n[:, TRIM_INICIO]
    bins = np.linspace(-40, 90, 66)
    for pagos, lab, col in [(r_n["pagos_sin"], "Sin cobertura", GR), (r_n["pagos_con"], "Cobertura rodante (enunciado)", AZ),
                            (rs_n["pagos_con"], "Strip fijado en el año 6", VE)]:
        rel = (pagos[:, 24:].sum(axis=1) / base24 - 1) * 100
        ax.hist(rel, bins=bins, histtype="step", lw=1.6, density=True, color=col,
                label=f"{lab}: desv. est. {_num(rel.std(), 1)} %")
    ax.set_title("(d) Costo de los años 7-10 frente a la TRM del año 6", fontsize=11)
    ax.set_xlabel("Sobrecosto"); ax.xaxis.set_major_formatter(pct); ax.set_yticks([])
    ax.set_ylim(0, ax.get_ylim()[1] * 1.35)
    ax.legend(fontsize=8, frameon=False, loc="upper right")

    # (e) Desviación de cada año frente a su presupuesto
    ax = fig.add_subplot(gs[1, 1])
    bins = np.linspace(-35, 35, 71)
    d_sin = (r_n["U"] / r_n["B"] - 1).ravel() * 100
    d_con = (r_n["H"] / r_n["B"] - 1).ravel() * 100
    ax.hist(d_sin, bins=bins, color=GR, alpha=0.55, density=True, label=f"Sin cobertura (> +10 %: {_num((d_sin > 10).mean()*100, 2)} %)")
    ax.hist(d_con, bins=bins, color=AZ, alpha=0.6, density=True, label=f"Con cobertura (> +10 %: {_num((d_con > 10).mean()*100, 2)} %)")
    ax.axvline(10, color=RO, ls="--", lw=1.2, label="Presupuesto + 10 % (umbral de fracaso)")
    ax.set_title("(e) Costo de cada año frente a su presupuesto al forward", fontsize=11)
    ax.set_xlabel("Desviación"); ax.xaxis.set_major_formatter(pct); ax.set_yticks([])
    ax.set_ylim(0, ax.get_ylim()[1] * 1.3)
    ax.legend(fontsize=8, frameon=False, loc="upper left")

    # (f) Sensibilidad a la prima forward
    ax = fig.add_subplot(gs[1, 2])
    s = sens[np.isclose(sens.i_USD, I_USD)].sort_values("prima_forward")
    ax.plot(s.prima_forward * 100, s.resultado_forwards_media / 1e6, "o-", color=AZ, label="Cobertura rodante")
    ax.plot(s.prima_forward * 100, s.strip_resultado_media / 1e6, "s-", color=VE, label="Strip")
    ax.axhline(0, color="#333", lw=0.8)
    ax.axvline(prima(I_COP, I_USD) * 100, color=RO, ls="--", lw=1.2, label=f"Prima de mercado ({_num(prima(I_COP, I_USD)*100, 1)} %)")
    ax.set_title(f"(f) Resultado esperado de los forwards según la prima\n(i_USD = {_num(I_USD*100, 2)} %)", fontsize=11)
    ax.set_xlabel("Prima forward anual"); ax.set_ylabel("Millones de COP")
    ax.xaxis.set_major_formatter(pct); ax.yaxis.set_major_formatter(fmt0); ax.legend(fontsize=8, frameon=False, loc="lower left")

    # (g) Escenarios representativos (normal)
    p = prima(I_COP, I_USD)
    q = np.arange(paths_n.shape[1])
    for k, nombre in enumerate(["Favorable", "Central", "Adverso"]):
        ax = fig.add_subplot(gs[2, k])
        S = rep[f"{nombre} – Normal"]
        ax.plot(q[20:], S[20:], color="#333", lw=1.6, label="TRM simulada")
        total = 0
        for i, (c, m) in enumerate(zip(PACTOS, VENCIMIENTOS)):
            Fk = S[c] * (1 + p)
            liq = COBERTURA * 4 * cuota * (S[m] - Fk)
            total += liq
            col = VE if S[m] > Fk else RO
            ax.hlines(Fk, c, m, color=col, lw=2.2, label="Forward pactado" if i == 0 else None)
            ax.plot([m, m], [Fk, S[m]], color=col, lw=1, ls=":")
            ax.annotate(_num(liq / 1e6, 1), (m, max(Fk, S[m])), textcoords="offset points", xytext=(-14, 6),
                        fontsize=8, color=col)
        ax.axvspan(24, 40, color="#f2f2f2", zorder=0)
        ax.set_title(f"(g) Escenario {nombre.lower()}: resultado de los forwards {_num(total / 1e6, 1)} millones", fontsize=10.5)
        ax.set_xlabel("Trimestre (zona gris: años cubiertos)"); ax.yaxis.set_major_formatter(fmt0)
        ax.set_xticks([20, 24, 28, 32, 36, 40])
        ax.set_ylim(ax.get_ylim()[0], ax.get_ylim()[1] + 0.06 * (ax.get_ylim()[1] - ax.get_ylim()[0]))
        if k == 0:
            ax.set_ylabel("COP por USD")
            ax.legend(fontsize=8, frameon=False, loc="upper left")

    for a in fig.axes:
        a.grid(alpha=0.25); a.spines[["top", "right"]].set_visible(False)
    fig.suptitle("Parte B: cobertura con cuatro forwards anuales (75 % de los pagos de los años 7 a 10)", fontsize=14, y=0.985)
    fig.savefig("graficas_cobertura_forward.png", dpi=150, bbox_inches="tight")
    plt.show()


if __name__ == "__main__":
    monto_usd, cuota, amort = sim.tabla_credito()
    desembolso = sim.VALOR_MAQUINARIA_COP * (1 - sim.APORTE)
    p = sim.estimar_parametros(sim.VENTANAS[sim.VENTANA_ELEGIDA])
    nu = sim.estimar_nu(sim.VENTANAS[sim.VENTANA_ELEGIDA])
    print(f"i_COP implícita {I_COP:.4%} | i_USD {I_USD:.4%} | prima forward {prima(I_COP, I_USD):.4%} | nu t {nu:.2f}")
    print(f"Paridad con IBR 12 meses ({I_COP_IBR:.4%} EA) y Tesoro ({I_USD:.4%} EA): forward teórico "
          f"{SPOT_MERCADO * (1 + I_COP_IBR) / (1 + I_USD):,.2f} vs observado {FWD_1Y_MERCADO:,.2f}")

    paths_n, _ = sim.simular(p["mu_q"], p["sig_q"])
    paths_t, _ = sim.simular(p["mu_q"], p["sig_q"], seed=sim.SEED + 1, dist="t", nu=nu)
    paths_t4, _ = sim.simular(p["mu_q"], p["sig_q"], seed=sim.SEED + 2, dist="t", nu=NU_STRESS)

    res_n, r_n = resumen(paths_n, cuota, "Normal", desembolso)
    res_t, r_t = resumen(paths_t, cuota, f"t de Student (nu = {nu:.1f})", desembolso)
    res_t4, r_t4 = resumen(paths_t4, cuota, f"t de Student de estrés (nu = {NU_STRESS})", desembolso)
    tabla = pd.concat([res_n, res_t, res_t4], axis=1)
    tabla.to_csv("cobertura_resultados.csv")
    with pd.option_context("display.width", 250, "display.max_rows", 200, "display.max_colwidth", 70,
                           "display.float_format", "{:,.4f}".format):
        t2 = tabla.copy(); t2.columns = ["Normal", "t", "t4"]
        print(t2.to_string())

    pf = pd.concat([por_forward(r_n).assign(dist="Normal"), por_forward(r_t).assign(dist="t")])
    pf.to_csv("cobertura_por_forward.csv", index=False)
    print(pf.round(3).to_string(index=False))

    # Trayectorias representativas: entre los caminos con TRM del año 6 cercana a la mediana
    # (percentiles 45-55), se toman los percentiles 10, 50 y 90 de la depreciación durante
    # el período cubierto (TRM trimestre 40 / TRM trimestre 24). Así los tres escenarios
    # parten del mismo nivel y difieren en lo que ocurre durante la cobertura.
    rep = {}
    for etiqueta, paths in [("Normal", paths_n), ("t", paths_t)]:
        s24 = paths[:, TRIM_INICIO]
        lo, hi = np.quantile(s24, [0.45, 0.55])
        cand = np.where((s24 >= lo) & (s24 <= hi))[0]
        dep = paths[cand, 40] / paths[cand, TRIM_INICIO]
        orden = cand[np.argsort(dep)]
        for nombre, q in [("Favorable", 0.10), ("Central", 0.50), ("Adverso", 0.90)]:
            j = orden[int(q * (len(orden) - 1))]
            rep[f"{nombre} – {etiqueta}"] = paths[j]
    pd.DataFrame(rep).to_csv("trayectorias_representativas.csv", index_label="trimestre")

    # Sensibilidad a tasas (innovaciones normales, mismas trayectorias)
    filas = []
    for i_cop in [0.06, 0.08, 0.10, 0.1225, I_COP_IBR, I_COP, 0.14]:
        for i_usd in [0.03, I_USD, 0.06]:
            r = evaluar(paths_n, cuota, i_cop, i_usd)
            rs = evaluar(paths_n, cuota, i_cop, i_usd, variante="strip")
            liq = r["liq"].sum(axis=1)
            filas.append({
                "i_COP": i_cop, "i_USD": i_usd, "prima_forward": prima(i_cop, i_usd),
                "prob_proteccion": (r["Sv"] > r["F"]).mean(),
                "resultado_forwards_media": liq.mean(),
                "costo_total_con_media": r["pagos_con"].sum(axis=1).mean(),
                "sd_costo_710_con": r["pagos_con"][:, 24:].sum(axis=1).std(),
                "prob_cobertura_cuesta_mas": (liq < 0).mean(),
                "strip_resultado_media": rs["liq"].sum(axis=1).mean(),
                "fracaso": (r["H"] > r["B"] * (1 + TOLERANCIA_PPTO)).mean(),
            })
    sens = pd.DataFrame(filas)
    sens.to_csv("sensibilidad_tasas.csv", index=False)
    print(sens.round(4).to_string(index=False))

    # Distribuciones para la figura de la parte B
    np.savez_compressed("cobertura_distribuciones.npz",
                        sin_n=r_n["pagos_sin"][:, 24:].sum(axis=1), con_n=r_n["pagos_con"][:, 24:].sum(axis=1),
                        sin_t=r_t["pagos_sin"][:, 24:].sum(axis=1), con_t=r_t["pagos_con"][:, 24:].sum(axis=1),
                        strip_n=evaluar(paths_n, cuota, variante="strip")["pagos_con"][:, 24:].sum(axis=1),
                        dev_sin=(r_n["U"] / r_n["B"] - 1).ravel(), dev_con=(r_n["H"] / r_n["B"] - 1).ravel())

    if MOSTRAR_GRAFICAS:
        graficas(paths_n, cuota, r_n, r_t, r_t4, evaluar(paths_n, cuota, variante="strip"), sens, rep, nu)
```
